## 1) Wahrscheinlichkeiten und Erwartungswert

Eine Tombola hat 100 Lose, jedes Los kostet 2 €. Es gibt folgende Gewinne: 
- 1 Hauptpreis: 50 €
- 4 Trostpreise: je 10 €
- der Rest: kein Gewinn

Bestimme den zu erwartenden Gewinn. Würdest du das Spiel spielen?

**Lösung:**

Im ersten Schritt werden die jeweiligen Wahrscheinlichkeiten bestimmt:  

| Gewinn | 50 EUR | 10 EUR | 0 EUR | 
| --- | --- | --- | --- | 
| $p$ | $0.01$ | $0.04$ | $0.95$ |
| Gewinn - Einsatz | 48 EUR | 8 EUR | -2 EUR | 

Nun können wir den zu erwartenden Gewinn bestimmen

$E(\text{Tombola Gewinn}) = 0.01\cdot 48 + 0.04\cdot 8 + 0.95 \cdot (-2)= 1.10$

```python
0.01*48 + 0.04*8 + 0.95*(-2)
```

Wenn du viele Lose kaufst, **verlierst** du im Durchschnitt 1,10 € pro Los.

## 2) Hähnchen oder doch Ratte? 

Ein Restaurant verkauft große Mengen an frittierten Hähnchenteilen. Aufgrund mangelnder
Hygiene bestehen die vermeintlichen Hähnchenteile mit einer Wahrscheinlichkeit von 2% aus
Ratte. Besteht ein verkauftes Teil aus Ratte, so fällt dieses zu 70% durch unangenehmen
Geschmack auf. Bei Hähnchen ist dies nur zu 5% der Fall.

a) Beim Essen fällt dir ein unangenehmer Geruch auf. Mit welcher Wahrscheinlichkeit befindet sich ein Rattenteilchen auf deinem Teller? **Interpretiere das Ergebnis.** 

b) Beim Essen fällt dir ein unangenehmer Geruch auf. Mit welcher Wahrscheinlichkeit befindet sich nur Hähnchen auf deinem Teller? **Interpretiere das Ergebnis.** 

**Lösung a)** 

\begin{eqnarray*}
P(R | u.G.) & = & \frac{ P(R)\cdot P( u.G. | R) }{P(R)\cdot P( u.G. | R) + P(H)\cdot P( u.G. | H)} \\
& = & \frac{ 0{,}02 \cdot 0{,}70 }{0{,}02 \cdot 0{,}70 + 0{,}98 \cdot 0{,}05} \\
& = & 22\%
\end{eqnarray*}

```python
( 2/100 * 70/100 ) / (2/100 * 70/100 + 98/100 * 5/100) 
```

**Lösung b)** 

\begin{eqnarray*}
P(H | u.G.) & = & \frac{ P(H)\cdot P( u.G. | H) }{P(R)\cdot P( u.G. | R) + P(H)\cdot P( u.G. | H)} \\
& = & \frac{ 0{,}98 \cdot 0{,}05 }{0{,}02 \cdot 0{,}70 + 0{,}98 \cdot 0{,}05} \\
& = & 78\%
\end{eqnarray*}

```python
( 98/100 * 5/100 ) / (2/100 * 70/100 + 98/100 * 5/100) 
```

## 3) Qualitätskontrolle #1
Eine Lieferung von 27 Stiften enthält 2 Stifte, die defekt sind. Der Empfänger testet zur Kontrolle 3 Stifte und behält die Lieferung, falls alle 3 funktionieren. 

a) Welche Verteilung liegt zugrunde. Warum? Visualiere pdf und cdf 

b) Wie groß ist die Wahrscheinlichkeit, dass 1 Stift defekt ist? 

c) Wie groß ist die Wahrscheinlichkeit, dass mindestens 1 Stift defekt ist? 

d) Wie groß ist die Wahrscheinlichkeit, dass der Empfänger die Lieferung behält? 

**Lösung a)** Es liegt eine Hypergeometrische Verteilung zugrunde, da NICHT zurückgelegt wird. --> Die Wahrscheinlichkeiten ändern sich nach jeder Inspektion. 

```python
#https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.hypergeom.html

# 27 Teile, 2 Defekt, 3 Stickprobe
[M, n, N] = [27, 2, 3]
rv = hypergeom(M, n, N)
x = np.arange(0, 3)
# Bestimme Wahrscheinlichkeiten
pmf_defStift = rv.pmf(x)
print(pmf_defStift) # Wahrscheinlichkeiten

# fig = plt.figure()
# ax = fig.add_subplot(111)
# plt.title('PDF - Wahrscheinlichkeitsverteilung')
# ax.plot(x, pmf_defStift, 'bo')
# ax.vlines(x, 0, pmf_defStift, lw=2)
# ax.set_xlabel('# defekter Stifte')
# ax.set_ylabel('Wahrscheinlichkeit')
# ax.xaxis.set_ticks([0,1,2])
# plt.show()
plt.stem(x, pmf_defStift)
plt.title('PDF - Wahrscheinlichkeitsverteilung')
plt.xlabel('# defekter Teile')
plt.ylabel('Wahrscheinlichkeit')
plt.xticks([0, 1, 2])
plt.show()

cdf_defStift = np.cumsum(pmf_defStift)
plt.step(x, cdf_defStift, where='post')
plt.title('CDF - kumulierte Verteilung')
plt.xlabel('# defekter Teile')
plt.ylabel('P(X ≤ k)')
plt.xticks([0, 1, 2])
plt.ylim(0, 1)
plt.show()
```

```python
# Verwende np.cumsum um kum. Wahrscheinlichkeiten (CDF) zu generieren

fig = plt.figure()
ax = fig.add_subplot(111)
plt.title('CDF - kumulative Verteilungsfunktion')
ax.plot(x, np.cumsum(pmf_defStift), 'bo')
ax.vlines(x, 0, np.cumsum(pmf_defStift), lw=2)
ax.set_xlabel('# defekter Stifte')
ax.set_ylabel('Kum. Wahrscheinlichkeit')
ax.xaxis.set_ticks([0,1,2])
plt.show()
```

## 4) Qualitätskontrolle #2

In einer Produktion ist der Anteil fehlerhafter Bauteile gleichbleibend $4\%$. $X = $ Anzahl an fehlerhaften Bauteilen. 

a) Welche Verteilung liegt zugrund? Warum? Visualisiere pdf und cdf. 

b) Wie groß ist die Wahrscheinlichkeit, dass in einer zufälligen Stichprobe von 500 Einheiten mindestens zwei und maximal fünf fehlerhafte Einheiten gefunden werden? 

c) Wieviele fehlerhafte Bauteile sind zu erwarten?

d) Wie groß muss die Stichprobe gewählt werden, um mit mindestens 90%iger Sicherheit
mindestens eine fehlerhafte Einheit zu entdecken?

e) Verwende bei der Aufgabenstellung b) zusätzlich die Approximation durch die Normalverteilung, falls erlaubt. Vergleiche die Ergebnisse 

**Lösung**

a) Eine Binomialverteilung liegt zugrund, weil die Wahrscheinlichkeit gleichbleibend ist.

```python
[n, p] = [500, 0.04]
rv = binom(n, p)
x = np.arange(0, n+1)

pmf_defekt = rv.pmf(x)
cdf_defekt = rv.cdf(x)

fig = plt.figure()
ax = fig.add_subplot(111)
plt.title('PDF - Wahrscheinlichkeitsverteilung')
ax.plot(x, pmf_defekt, 'bo')
ax.vlines(x, 0, pmf_defekt, lw=2)
ax.set_xlabel('# fehlerhafte Bauteile')
ax.set_ylabel('Wahrscheinlichkeit')
plt.show()

fig = plt.figure()
ax = fig.add_subplot(111)
plt.title('CDF - Kum. Verteilungsfunktion')
ax.plot(x, cdf_defekt, 'bo')
ax.vlines(x, 0, cdf_defekt, lw=2)
ax.set_xlabel('# fehlerhafte Bauteile')
ax.set_ylabel('Kum. Wahrscheinlichkeit')
plt.show()
```

b) Wie groß ist die Wahrscheinlichkeit, dass in einer zufälligen Stichprobe von 500 Einheiten mindestens 2 und maximal 5 fehlerhafte Einheit gefunden werden?

$P(2 \leq X \leq 5) = \binom{500}{2}\cdot 0,04^{2}\cdot 0,96^{498} + \binom{500}{3}\cdot 0,04^{3}\cdot 0,96^{497} + ... + \binom{500}{5}\cdot 0,04^{5}\cdot 0,96^{495} =  0{,}0057\%$

```python
## in python:

x = np.arange(2, 6) # 6 exclusive!
print(x)
res = sum(rv.pmf(x))

print('Die Wahrscheinlichkeit beträgt {:.4f}%'.format(res*100))
```

c) Wieviele fehlerhafte Bauteile sind zu erwarten?

$E(X) = n\cdot p = 500\cdot 0.04 = 20$

```python
500*4/100
```

d) Wie groß muss die Stichprobe gewählt werden, um mit mindestens 90%iger Sicherheit
mindestens eine fehlerhafte Einheit zu entdecken?

**Exakt über Formel:**

$P(X \geq 1) \geq 0.9$

$1 - P(X = 0) \geq 0.9$

$1 - 0.9  \geq P(X = 0)$

$0.1  \geq P(X = 0) = \binom{n}{0}\cdot 0.04^{0}\cdot 0.96^{n}$

$0.1  \geq  0.96^{n}$

$\frac{ln(0.1)}{ln(0.96)}  \leq  n$

$56.4  \leq  n$

$\Rightarrow$ Die Stichprobe muss mindestens 57 Teile beinhalten. 

e) Verwende bei den Aufgabenstellungen b) zusätzlich die Approximation durch die Normalverteilung, falls erlaubt. Vergleiche die Ergebnisse

Bedingung $n\cdot p \cdot (1-p) = 19,2$. Es darf somit die Normalverteilung verwendet werden mit $\mu = n\cdot p = 20$ und $\sigma^2 = n\cdot p \cdot (1-p) = 19,2$

**Achtung:** Stetigkeitskorrektur!

```python
loc = 20
scale = np.sqrt(19.2)

res = norm.cdf(5.5,loc=loc,scale=scale) - norm.cdf(2.5,loc=loc,scale=scale) 
print('Die Wahrscheinlichkeit beträgt {:.4f}%'.format(res*100))
```

Vergleiche die Ergebnisse: Obwohl die Approximation theoretisch OK ist, ist der Fehler trotzdem noch groß (~10-fache).

## 6) Konfidenzintervalle

Bestimme die Konfidenzintervalle für den Erwartungswert von Flipper Length (mm) der verschiedenen Pinguine zum Niveau $\alpha = 0.05$. 

```python
penguins = pd.read_excel('./data/penguinsSmall.xlsx', index_col=0)

penguins.head()

Adelie_Flipper_Length = penguins.loc[penguins['Species'] == "Adelie", ['Flipper Length (mm)']]
Chinstrap_Flipper_Length = penguins.loc[penguins['Species'] == "Chinstrap", ['Flipper Length (mm)']]
Gentoo_Flipper_Length = penguins.loc[penguins['Species'] == "Gentoo", ['Flipper Length (mm)']]
```

```python
mean_Adelie = Adelie_Flipper_Length.mean()[0]
mean_Chinstrap = Chinstrap_Flipper_Length.mean()[0]
mean_Gentoo = Gentoo_Flipper_Length.mean()[0]
```

```python
# define ddof = 1 => it is divided by (n-1), for the sample standard deviation
std_Adelie = np.std(Adelie_Flipper_Length, ddof = 1)[0]
std_Chinstrap = np.std(Chinstrap_Flipper_Length, ddof = 1)[0]
std_Gentoo = np.std(Gentoo_Flipper_Length, ddof = 1)[0]
```

```python
# t-statistik
alpha = 0.05
```

### Adelie

```python
n = Adelie_Flipper_Length.count()[0]
t_stats = round(t.ppf(1-alpha/2 , n-1), 3)

LB = round(mean_Adelie - t_stats*std_Adelie/np.sqrt(n) , 2)
UB = round(mean_Adelie + t_stats*std_Adelie/np.sqrt(n) , 2)

print('Adelie: Konfidenzintervall für den Erwartungswert: [{0} ; {1}]'.format(LB, UB))
```

### Chinstrap

```python
n = Chinstrap_Flipper_Length.count()[0]
t_stats = round(t.ppf(1-alpha/2 , n-1), 3)

LB = round(mean_Chinstrap - t_stats*std_Chinstrap/np.sqrt(n) , 2)
UB = round(mean_Chinstrap + t_stats*std_Chinstrap/np.sqrt(n) , 2)

print('Chinstrap: Konfidenzintervall für den Erwartungswert: [{0} ; {1}]'.format(LB, UB))
```

### Gentoo

```python
n = Gentoo_Flipper_Length.count()[0]
t_stats = round(t.ppf(1-alpha/2 , n-1), 3)

LB = round(mean_Gentoo - t_stats*std_Gentoo/np.sqrt(n) , 2)
UB = round(mean_Gentoo + t_stats*std_Gentoo/np.sqrt(n) , 2)

print('Chinstrap: Konfidenzintervall für den Erwartungswert: [{0} ; {1}]'.format(LB, UB))
```

## 7) Hypothesentests

Ein Tierarzt behauptet, dass der Erwartungswert der Flipper Length (mm) der unterschiedlichen Pinguine 190 misst. Überprüfe diese Aussage mittels Hypothesentest (Signifikanzniveau  𝛼=0.05 ). 

```python
# we need an array for the ttest()
Adelie_Flipper_Length_array = np.array(Adelie_Flipper_Length.iloc[:,0])
Chinstrap_Flipper_Length_array = np.array(Chinstrap_Flipper_Length.iloc[:,0])
Gentoo_Flipper_Length_array = np.array(Gentoo_Flipper_Length.iloc[:,0])
```

```python
result = pg.ttest(Adelie_Flipper_Length_array,
                  190)
result
```
