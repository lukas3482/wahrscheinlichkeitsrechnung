## Aufgabe 4: $\chi^2$-Verteilung


Visualisiere pdf und cdf einer $\chi^2$-Verteilung mit Freiheitsgraden $\nu \in  [2, 3, 5, 10, 20, 30]$ auf dem Intevall $0 \leq x \leq 50$. 

Berechne jeweils Erwartungswert, Standardabweichung, Schiefe und Kurtosis. Was fällt auf, wenn du es mit einer Normalverteilung vergleichst? 

```python
# Freiheitsgrade
dfs = [2,3,5,10,20,30]
x = np.arange(0, 50)

for df in dfs:
    plt.plot(x, chi2.pdf(x, df=df), label=df)

plt.legend()
plt.show()

for df in dfs:
    plt.plot(x, chi2.cdf(x, df=df), label=df)

plt.legend()
plt.show()

for df in dfs:
    mean, var, skew, kurt = chi2.stats(df, moments='mvsk')
    print("Freiheitsgrade: ", df)
    print("Mean: {}, Var: {}, Skew: {}, Kurt: {}".format(mean, var, round(skew, 4), kurt))

```
